```python
import os
import json
import pickle
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import h5py
import yaml

from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

warnings.filterwarnings("ignore")

# ============================================================
# 1. FILE PATH CONFIGURATION
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Electricity Consumption Pattern Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "Consumptionofelectricity_0.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

print("=" * 75)
print("AI-BASED ELECTRICITY CONSUMPTION PATTERN ANALYSIS")
print("=" * 75)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH, encoding="utf-8-sig")

print("\nDataset loaded successfully!")
print("\nDataset shape:", df.shape)
print("\nOriginal columns:")
print(df.columns.tolist())

print("\nFirst five records:")
print(df.head())

# ============================================================
# 3. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
)

column_mapping = {}

for col in df.columns:

    normalized = (
        col.lower()
        .replace(" ", "")
        .replace("_", "")
    )

    if "category" in normalized or "sector" in normalized:
        column_mapping[col] = "Category"

    elif "percentage" in normalized:
        column_mapping[col] = "Percentage"

    elif "consumption" in normalized:
        column_mapping[col] = "Consumption"

df.rename(columns=column_mapping, inplace=True)

required_columns = ["Category", "Consumption"]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

df["Category"] = (
    df["Category"]
    .astype(str)
    .str.strip()
)

df["Consumption"] = pd.to_numeric(
    df["Consumption"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace(
        r"[^0-9.\-]",
        "",
        regex=True
    ),
    errors="coerce"
)

df = df.replace([np.inf, -np.inf], np.nan)

df = df.dropna(
    subset=["Category", "Consumption"]
)

df = df[
    (df["Category"] != "") &
    (df["Category"].str.lower() != "nan") &
    (df["Consumption"] >= 0)
].copy()

# Remove aggregate rows if present.
# They are not individual electricity-consuming sectors.
aggregate_names = {
    "total",
    "grand total",
    "all sectors",
    "overall total"
}

df = df[
    ~df["Category"].str.lower().isin(
        aggregate_names
    )
].copy()

# Combine duplicate sector names if any
df = (
    df.groupby("Category", as_index=False)
    ["Consumption"]
    .sum()
)

if len(df) < 3:
    raise ValueError(
        "At least three valid sector records are required."
    )

# ============================================================
# 5. FEATURE ENGINEERING
# ============================================================

total_consumption = df["Consumption"].sum()

if total_consumption <= 0:
    raise ValueError(
        "Total electricity consumption must be positive."
    )

df["Percentage"] = (
    df["Consumption"] / total_consumption
) * 100

df["Consumption_Share"] = (
    df["Consumption"] / total_consumption
)

df["Consumption_Rank"] = (
    df["Consumption"]
    .rank(
        ascending=False,
        method="dense"
    )
    .astype(int)
)

df["Log_Consumption"] = np.log1p(
    df["Consumption"]
)

df["Normalized_Consumption"] = (
    df["Consumption"] /
    df["Consumption"].max()
)

# ============================================================
# 6. CREATE CLASSIFICATION LABELS
# ============================================================

# Tertile-based classification:
# Low    = lower consumption group
# Medium = middle consumption group
# High   = higher consumption group
#
# These are relative labels for this dataset,
# not official energy-demand standards.

lower_threshold = float(
    df["Consumption"].quantile(1 / 3)
)

upper_threshold = float(
    df["Consumption"].quantile(2 / 3)
)

class_names = ["Low", "Medium", "High"]

class RuleBasedConsumptionClassifier(
    BaseEstimator,
    ClassifierMixin
):
    def __init__(
        self,
        low_threshold=0.0,
        high_threshold=1.0
    ):
        self.low_threshold = low_threshold
        self.high_threshold = high_threshold

    def fit(self, X, y=None):
        self.classes_ = np.array(
            ["Low", "Medium", "High"]
        )
        self.n_features_in_ = 1
        return self

    def predict(self, X):
        X = np.asarray(X, dtype=float)

        # Input is the standardized consumption
        # feature from the pipeline.
        values = X[:, 0]

        return np.where(
            values <= self.low_threshold,
            "Low",
            np.where(
                values <= self.high_threshold,
                "Medium",
                "High"
            )
        )

# ============================================================
# 7. BUILD CLASSIFICATION PIPELINE
# ============================================================

X = df[["Consumption"]].values

scaler = StandardScaler()
scaler.fit(X)

# Convert raw thresholds into standardized units
scaled_lower = (
    lower_threshold - scaler.mean_[0]
) / scaler.scale_[0]

scaled_upper = (
    upper_threshold - scaler.mean_[0]
) / scaler.scale_[0]

model = Pipeline(
    steps=[
        ("scaler", StandardScaler()),
        (
            "classifier",
            RuleBasedConsumptionClassifier(
                low_threshold=float(scaled_lower),
                high_threshold=float(scaled_upper)
            )
        )
    ]
)

model.fit(X)

df["Consumption_Level"] = model.predict(X)

label_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

df["Consumption_Level_Encoded"] = (
    df["Consumption_Level"]
    .map(label_mapping)
    .astype(int)
)

print("\nClassification completed successfully!")

print("\nSector-wise classification:")
print(
    df[
        [
            "Category",
            "Consumption",
            "Percentage",
            "Consumption_Level"
        ]
    ].to_string(index=False)
)

# ============================================================
# 8. CALCULATE DATASET STATISTICS
# ============================================================

statistics = {
    "total_sectors": int(len(df)),
    "total_consumption_MU": float(
        total_consumption
    ),
    "average_consumption_MU": float(
        df["Consumption"].mean()
    ),
    "median_consumption_MU": float(
        df["Consumption"].median()
    ),
    "minimum_consumption_MU": float(
        df["Consumption"].min()
    ),
    "maximum_consumption_MU": float(
        df["Consumption"].max()
    ),
    "standard_deviation_MU": float(
        df["Consumption"].std(ddof=0)
    ),
    "highest_consumption_sector": str(
        df.loc[
            df["Consumption"].idxmax(),
            "Category"
        ]
    ),
    "lowest_consumption_sector": str(
        df.loc[
            df["Consumption"].idxmin(),
            "Category"
        ]
    )
}

class_distribution = {
    name: int(
        (df["Consumption_Level"] == name).sum()
    )
    for name in class_names
}

print("\nDataset statistics:")

for key, value in statistics.items():
    print(f"{key}: {value}")

print("\nClass distribution:")
print(class_distribution)

# ============================================================
# 9. SAVE TRAINED PIPELINE AS PKL
# ============================================================

pkl_path = os.path.join(
    BASE_DIR,
    "electricity_consumption_model.pkl"
)

with open(pkl_path, "wb") as file:
    pickle.dump(model, file)

print("\nPKL model saved:", pkl_path)

# ============================================================
# 10. SAVE MODEL PARAMETERS AND DATA AS H5
# ============================================================

h5_path = os.path.join(
    BASE_DIR,
    "electricity_consumption_model.h5"
)

with h5py.File(h5_path, "w") as h5file:

    h5file.attrs["project_name"] = (
        "AI-Based Electricity Consumption "
        "Pattern Analysis"
    )

    h5file.attrs["model_type"] = (
        "Rule-Based Tertile Classification"
    )

    h5file.attrs["feature"] = "Consumption"

    h5file.attrs["created_at"] = (
        datetime.now().isoformat()
    )

    # Model parameters
    parameters = h5file.create_group(
        "model_parameters"
    )

    parameters.create_dataset(
        "lower_threshold",
        data=lower_threshold
    )

    parameters.create_dataset(
        "upper_threshold",
        data=upper_threshold
    )

    fitted_scaler = model.named_steps["scaler"]

    parameters.create_dataset(
        "scaler_mean",
        data=fitted_scaler.mean_
    )

    parameters.create_dataset(
        "scaler_scale",
        data=fitted_scaler.scale_
    )

    # Dataset
    data_group = h5file.create_group(
        "processed_data"
    )

    string_dtype = h5py.string_dtype(
        encoding="utf-8"
    )

    data_group.create_dataset(
        "categories",
        data=df["Category"].tolist(),
        dtype=string_dtype
    )

    data_group.create_dataset(
        "consumption",
        data=df["Consumption"].to_numpy()
    )

    data_group.create_dataset(
        "percentage",
        data=df["Percentage"].to_numpy()
    )

    data_group.create_dataset(
        "consumption_levels",
        data=df["Consumption_Level"].tolist(),
        dtype=string_dtype
    )

    data_group.create_dataset(
        "encoded_labels",
        data=df[
            "Consumption_Level_Encoded"
        ].to_numpy()
    )

print("H5 file saved:", h5_path)

# ============================================================
# 11. SAVE YAML CONFIGURATION
# ============================================================

yaml_path = os.path.join(
    BASE_DIR,
    "electricity_consumption_config.yaml"
)

config = {
    "project": {
        "name": (
            "AI-Based Electricity Consumption "
            "Pattern Analysis"
        ),
        "description": (
            "Sector-wise electricity consumption "
            "analysis and classification"
        ),
        "dataset": DATASET_PATH,
        "output_directory": BASE_DIR
    },

    "model": {
        "type": "RuleBasedConsumptionClassifier",
        "classification_method": "tertiles",
        "feature_columns": ["Consumption"],
        "target_column": "Consumption_Level",
        "classes": class_names,
        "class_mapping": label_mapping,
        "lower_threshold_MU": lower_threshold,
        "upper_threshold_MU": upper_threshold
    },

    "preprocessing": {
        "missing_values": "remove_invalid_records",
        "duplicate_categories": "aggregate_sum",
        "scaling": "StandardScaler"
    },

    "dataset_information": {
        "records": int(len(df)),
        "features": int(len(df.columns)),
        "total_consumption_MU": float(
            total_consumption
        )
    },

    "output_files": {
        "h5": os.path.basename(h5_path),
        "pkl": os.path.basename(pkl_path),
        "yaml": os.path.basename(yaml_path),
        "json": "electricity_consumption_results.json"
    }
}

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("YAML configuration saved:", yaml_path)

# ============================================================
# 12. SAVE JSON RESULTS
# ============================================================

json_path = os.path.join(
    BASE_DIR,
    "electricity_consumption_results.json"
)

sector_results = []

for _, row in df.iterrows():

    sector_results.append({
        "category": str(row["Category"]),
        "consumption_MU": float(
            row["Consumption"]
        ),
        "percentage": float(
            row["Percentage"]
        ),
        "consumption_rank": int(
            row["Consumption_Rank"]
        ),
        "consumption_level": str(
            row["Consumption_Level"]
        ),
        "encoded_level": int(
            row["Consumption_Level_Encoded"]
        )
    })

json_results = {
    "project_name": (
        "AI-Based Electricity Consumption "
        "Pattern Analysis"
    ),

    "model_type": (
        "Rule-Based Tertile Classification"
    ),

    "generated_at": datetime.now().isoformat(),

    "statistics": statistics,

    "classification_thresholds_MU": {
        "low_upper_bound": lower_threshold,
        "medium_upper_bound": upper_threshold
    },

    "class_distribution": class_distribution,

    "sector_results": sector_results
}

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_results,
        file,
        indent=4,
        ensure_ascii=False
    )

print("JSON results saved:", json_path)

# ============================================================
# 13. SAVE PROCESSED DATASET
# ============================================================

csv_path = os.path.join(
    BASE_DIR,
    "processed_electricity_consumption.csv"
)

df.to_csv(
    csv_path,
    index=False,
    encoding="utf-8-sig"
)

print("Processed CSV saved:", csv_path)

# ============================================================
# 14. VERIFY SAVED PKL MODEL
# ============================================================

with open(pkl_path, "rb") as file:
    loaded_model = pickle.load(file)

test_predictions = loaded_model.predict(
    df[["Consumption"]].values
)

assert np.array_equal(
    test_predictions,
    df["Consumption_Level"].values
)

print("\nPKL model verification successful!")

# ============================================================
# 15. VERIFY H5 CONTENTS
# ============================================================

with h5py.File(h5_path, "r") as file:

    print("\nH5 file contents:")

    def show_h5_items(name, obj):
        print(name)

    file.visititems(show_h5_items)

# ============================================================
# 16. FINAL OUTPUT SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("PROJECT EXECUTION COMPLETED SUCCESSFULLY")
print("=" * 75)

output_files = [
    h5_path,
    pkl_path,
    yaml_path,
    json_path,
    csv_path
]

for path in output_files:

    print(
        f"{os.path.basename(path)} "
        f"| Exists: {os.path.exists(path)}"
    )

print("\nAll files saved in:")
print(BASE_DIR)

print("\nNOTE:")
print(
    "Classification is based on relative consumption "
    "thresholds, not a supervised ML accuracy score."
)
```
