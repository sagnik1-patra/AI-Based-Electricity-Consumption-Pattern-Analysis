```python
import os
import json
import pickle
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.pipeline import Pipeline

warnings.filterwarnings("ignore")

# ================================================================
# 1. PROJECT CONFIGURATION
# ================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Electricity Consumption Pattern Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "Consumptionofelectricity_0.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

CLASS_NAMES = ["Low", "Medium", "High"]

CLASS_COLORS = {
    "Low": "#2ecc71",
    "Medium": "#f39c12",
    "High": "#e74c3c"
}

print("=" * 75)
print("AI-BASED ELECTRICITY CONSUMPTION PATTERN ANALYSIS")
print("=" * 75)

# ================================================================
# 2. LOAD DATASET
# ================================================================

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nOriginal columns:")
print(df.columns.tolist())

print("\nFirst five rows:")
print(df.head())

# ================================================================
# 3. CLEAN COLUMN NAMES
# ================================================================

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
)

rename_columns = {}

for col in df.columns:

    normalized = (
        col.lower()
        .replace(" ", "")
        .replace("_", "")
    )

    if "category" in normalized or "sector" in normalized:
        rename_columns[col] = "Category"

    elif "percentage" in normalized:
        rename_columns[col] = "Percentage"

    elif "consumption" in normalized:
        rename_columns[col] = "Consumption"

df.rename(
    columns=rename_columns,
    inplace=True
)

required_columns = [
    "Category",
    "Consumption"
]

for col in required_columns:

    if col not in df.columns:
        raise ValueError(
            f"Required column missing: {col}"
        )

# ================================================================
# 4. DATA PREPROCESSING
# ================================================================

df["Category"] = (
    df["Category"]
    .astype(str)
    .str.strip()
)

df["Consumption"] = pd.to_numeric(
    df["Consumption"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace(
        r"[^0-9.\-]",
        "",
        regex=True
    ),
    errors="coerce"
)

df.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)

df.dropna(
    subset=["Category", "Consumption"],
    inplace=True
)

df = df[
    (df["Category"] != "") &
    (df["Category"].str.lower() != "nan") &
    (df["Consumption"] >= 0)
].copy()

# Remove total rows if present
aggregate_names = {
    "total",
    "grand total",
    "all sectors",
    "overall total"
}

df = df[
    ~df["Category"]
    .str.lower()
    .isin(aggregate_names)
].copy()

# Aggregate duplicate sector names
df = (
    df.groupby(
        "Category",
        as_index=False
    )["Consumption"]
    .sum()
)

if len(df) < 3:
    raise ValueError(
        "At least 3 valid sectors are required."
    )

df = df.sort_values(
    "Consumption",
    ascending=False
).reset_index(drop=True)

print("\nCleaned dataset:")
print(df)

# ================================================================
# 5. FEATURE ENGINEERING
# ================================================================

total_consumption = df["Consumption"].sum()

if total_consumption <= 0:
    raise ValueError(
        "Total consumption must be greater than zero."
    )

df["Percentage"] = (
    df["Consumption"] /
    total_consumption
) * 100

df["Consumption_Share"] = (
    df["Consumption"] /
    total_consumption
)

df["Normalized_Consumption"] = (
    df["Consumption"] /
    df["Consumption"].max()
)

df["Log_Consumption"] = np.log1p(
    df["Consumption"]
)

df["Consumption_Rank"] = (
    df["Consumption"]
    .rank(
        ascending=False,
        method="dense"
    )
    .astype(int)
)

# ================================================================
# 6. CREATE CONSUMPTION CLASSIFICATION
# ================================================================

# Relative tertile thresholds
lower_threshold = float(
    df["Consumption"].quantile(1 / 3)
)

upper_threshold = float(
    df["Consumption"].quantile(2 / 3)
)

def classify_consumption(value):

    if value <= lower_threshold:
        return "Low"

    elif value <= upper_threshold:
        return "Medium"

    return "High"

df["Actual_Class"] = (
    df["Consumption"]
    .apply(classify_consumption)
)

label_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

df["Actual_Encoded"] = (
    df["Actual_Class"]
    .map(label_mapping)
    .astype(int)
)

# ================================================================
# 7. BUILD REUSABLE CLASSIFICATION MODEL
# ================================================================

class ElectricityConsumptionClassifier(
    BaseEstimator,
    ClassifierMixin
):

    def __init__(
        self,
        lower_threshold=0.0,
        upper_threshold=1.0
    ):

        self.lower_threshold = lower_threshold
        self.upper_threshold = upper_threshold

    def fit(self, X, y=None):

        self.classes_ = np.array(CLASS_NAMES)
        self.n_features_in_ = 1

        return self

    def predict(self, X):

        X = np.asarray(X, dtype=float)

        values = X[:, 0]

        predictions = np.where(
            values <= self.lower_threshold,
            "Low",
            np.where(
                values <= self.upper_threshold,
                "Medium",
                "High"
            )
        )

        return predictions

X = df[["Consumption"]].values

scaler = StandardScaler()
scaler.fit(X)

scaled_lower = (
    lower_threshold - scaler.mean_[0]
) / scaler.scale_[0]

scaled_upper = (
    upper_threshold - scaler.mean_[0]
) / scaler.scale_[0]

model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        ElectricityConsumptionClassifier(
            lower_threshold=float(scaled_lower),
            upper_threshold=float(scaled_upper)
        )
    )
])

model.fit(X)

df["Predicted_Class"] = model.predict(X)

df["Predicted_Encoded"] = (
    df["Predicted_Class"]
    .map(label_mapping)
    .astype(int)
)

df["Correct_Classification"] = (
    df["Actual_Class"] ==
    df["Predicted_Class"]
)

# ================================================================
# 8. DESCRIPTIVE CLASSIFICATION AGREEMENT
# ================================================================

# This measures agreement with the same rules used
# to generate the labels.
#
# It is NOT independent test-set accuracy.

agreement = (
    df["Correct_Classification"].mean()
) * 100

class_agreement = {}

for class_name in CLASS_NAMES:

    subset = df[
        df["Actual_Class"] == class_name
    ]

    if len(subset) > 0:

        class_agreement[class_name] = float(
            subset["Correct_Classification"].mean()
            * 100
        )

    else:
        class_agreement[class_name] = 0.0

print("\nClassification agreement:", agreement)

print(
    "\nNOTE: This is rule agreement, "
    "not supervised ML test accuracy."
)

# ================================================================
# 9. ACCURACY GRAPH
# ================================================================

# Requested filename: accuracy_graph.png
# Chart explicitly identifies the metric as
# rule-based agreement to avoid misleading claims.

plt.figure(figsize=(10, 6))

categories = CLASS_NAMES + ["Overall"]

values = [
    class_agreement["Low"],
    class_agreement["Medium"],
    class_agreement["High"],
    agreement
]

colors = [
    CLASS_COLORS["Low"],
    CLASS_COLORS["Medium"],
    CLASS_COLORS["High"],
    "#3498db"
]

bars = plt.bar(
    categories,
    values,
    color=colors,
    edgecolor="black",
    alpha=0.85
)

for bar, value in zip(bars, values):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 1,
        f"{value:.2f}%",
        ha="center",
        fontweight="bold"
    )

plt.title(
    "Rule-Based Classification Agreement\n"
    "(Not Independent Test Accuracy)",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Consumption Class")
plt.ylabel("Agreement (%)")

plt.ylim(0, 110)

plt.tight_layout()

accuracy_path = os.path.join(
    BASE_DIR,
    "accuracy_graph.png"
)

plt.savefig(
    accuracy_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", accuracy_path)

# ================================================================
# 10. CORRELATION HEATMAP
# ================================================================

numeric_columns = [
    "Consumption",
    "Percentage",
    "Consumption_Share",
    "Normalized_Consumption",
    "Log_Consumption",
    "Consumption_Rank"
]

correlation_matrix = (
    df[numeric_columns]
    .corr()
)

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    vmin=-1,
    vmax=1
)

plt.title(
    "Electricity Consumption Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

heatmap_path = os.path.join(
    BASE_DIR,
    "heatmap.png"
)

plt.savefig(
    heatmap_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", heatmap_path)

# ================================================================
# 11. COMPARISON GRAPH
# ================================================================

# Compare sector consumption against
# the overall mean consumption.

average_consumption = df["Consumption"].mean()

x_positions = np.arange(len(df))

bar_width = 0.60

plt.figure(figsize=(13, 7))

bars = plt.bar(
    x_positions,
    df["Consumption"],
    width=bar_width,
    color=[
        CLASS_COLORS[level]
        for level in df["Actual_Class"]
    ],
    edgecolor="black"
)

plt.axhline(
    average_consumption,
    color="blue",
    linestyle="--",
    linewidth=2,
    label=(
        f"Average Consumption: "
        f"{average_consumption:,.2f} M.U."
    )
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

plt.xticks(
    x_positions,
    df["Category"],
    rotation=35,
    ha="right"
)

plt.xlabel("Electricity Consumption Sector")
plt.ylabel("Consumption (Million Units)")

plt.title(
    "Sector-Wise Electricity Consumption Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.legend()

plt.tight_layout()

comparison_path = os.path.join(
    BASE_DIR,
    "comparison_graph.png"
)

plt.savefig(
    comparison_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", comparison_path)

# ================================================================
# 12. GENERATE RESULT CSV
# ================================================================

result_df = df[
    [
        "Category",
        "Consumption",
        "Percentage",
        "Consumption_Rank",
        "Actual_Class",
        "Predicted_Class",
        "Correct_Classification"
    ]
].copy()

result_path = os.path.join(
    BASE_DIR,
    "result.csv"
)

result_df.to_csv(
    result_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nResult CSV generated:")
print(result_df)

print("Saved:", result_path)

# ================================================================
# 13. RESULT GRAPH
# ================================================================

# Show consumption share by classification level.

result_summary = (
    df.groupby("Actual_Class")["Consumption"]
    .sum()
    .reindex(CLASS_NAMES, fill_value=0)
)

plt.figure(figsize=(10, 7))

bars = plt.bar(
    result_summary.index,
    result_summary.values,
    color=[
        CLASS_COLORS[name]
        for name in result_summary.index
    ],
    edgecolor="black"
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.xlabel("Consumption Classification")
plt.ylabel("Total Electricity Consumption (M.U.)")

plt.title(
    "Electricity Consumption Classification Results",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

result_graph_path = os.path.join(
    BASE_DIR,
    "result_graph.png"
)

plt.savefig(
    result_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", result_graph_path)

# ================================================================
# 14. GENERATE PREDICTION CSV
# ================================================================

# These are classifications of existing sector data,
# not forecasts of future electricity demand.

prediction_df = pd.DataFrame({

    "Category": df["Category"],

    "Actual_Consumption_MU": (
        df["Consumption"]
    ),

    "Predicted_Consumption_Level": (
        df["Predicted_Class"]
    ),

    "Predicted_Class_Code": (
        df["Predicted_Encoded"]
    ),

    "Consumption_Percentage": (
        df["Percentage"]
    )

})

prediction_path = os.path.join(
    BASE_DIR,
    "prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nPrediction CSV generated:")
print(prediction_df)

print("Saved:", prediction_path)

# ================================================================
# 15. PREDICTION GRAPH
# ================================================================

plt.figure(figsize=(13, 7))

x_positions = np.arange(len(prediction_df))

plt.bar(
    x_positions,
    prediction_df["Actual_Consumption_MU"],
    color=[
        CLASS_COLORS[level]
        for level in
        prediction_df["Predicted_Consumption_Level"]
    ],
    edgecolor="black",
    alpha=0.85
)

for i, row in prediction_df.iterrows():

    plt.text(
        i,
        row["Actual_Consumption_MU"],
        row["Predicted_Consumption_Level"],
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold"
    )

plt.xticks(
    x_positions,
    prediction_df["Category"],
    rotation=35,
    ha="right"
)

plt.xlabel("Electricity Consumption Sector")
plt.ylabel("Consumption (Million Units)")

plt.title(
    "Predicted Electricity Consumption Level by Sector",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

prediction_graph_path = os.path.join(
    BASE_DIR,
    "prediction_graph.png"
)

plt.savefig(
    prediction_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", prediction_graph_path)

# ================================================================
# 16. CONFUSION MATRIX
# ================================================================

# Descriptive agreement matrix only.
# Not an independent model evaluation.

cm = confusion_matrix(
    df["Actual_Class"],
    df["Predicted_Class"],
    labels=CLASS_NAMES
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.xlabel("Predicted Classification")
plt.ylabel("Rule-Generated Classification")

plt.title(
    "Rule-Based Classification Agreement Matrix"
)

plt.tight_layout()

confusion_path = os.path.join(
    BASE_DIR,
    "confusion_matrix.png"
)

plt.savefig(
    confusion_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", confusion_path)

# ================================================================
# 17. SAVE MODEL AS PKL
# ================================================================

pkl_path = os.path.join(
    BASE_DIR,
    "electricity_consumption_model.pkl"
)

with open(pkl_path, "wb") as file:

    pickle.dump(model, file)

print("\nPKL model saved:", pkl_path)

# ================================================================
# 18. SAVE H5 MODEL PARAMETERS
# ================================================================

h5_path = os.path.join(
    BASE_DIR,
    "electricity_consumption_model.h5"
)

with h5py.File(h5_path, "w") as h5file:

    h5file.attrs["model_type"] = (
        "Rule-Based Electricity Classification"
    )

    h5file.attrs["created_at"] = (
        datetime.now().isoformat()
    )

    parameters = h5file.create_group(
        "model_parameters"
    )

    parameters.create_dataset(
        "lower_threshold",
        data=lower_threshold
    )

    parameters.create_dataset(
        "upper_threshold",
        data=upper_threshold
    )

    fitted_scaler = model.named_steps["scaler"]

    parameters.create_dataset(
        "scaler_mean",
        data=fitted_scaler.mean_
    )

    parameters.create_dataset(
        "scaler_scale",
        data=fitted_scaler.scale_
    )

    string_dtype = h5py.string_dtype(
        encoding="utf-8"
    )

    data_group = h5file.create_group(
        "processed_data"
    )

    data_group.create_dataset(
        "categories",
        data=df["Category"].tolist(),
        dtype=string_dtype
    )

    data_group.create_dataset(
        "consumption",
        data=df["Consumption"].to_numpy()
    )

    data_group.create_dataset(
        "predicted_classes",
        data=df["Predicted_Class"].tolist(),
        dtype=string_dtype
    )

print("H5 model parameters saved:", h5_path)

# ================================================================
# 19. SAVE YAML CONFIGURATION
# ================================================================

yaml_path = os.path.join(
    BASE_DIR,
    "electricity_consumption_config.yaml"
)

configuration = {

    "project_name": (
        "AI-Based Electricity Consumption "
        "Pattern Analysis"
    ),

    "dataset": DATASET_PATH,

    "model_type": (
        "Rule-Based Tertile Classification"
    ),

    "classification": {
        "classes": CLASS_NAMES,
        "lower_threshold_MU": lower_threshold,
        "upper_threshold_MU": upper_threshold
    },

    "evaluation": {
        "metric": "rule_agreement",
        "agreement_percentage": float(agreement),
        "independent_test_accuracy": None
    },

    "output_directory": BASE_DIR
}

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        configuration,
        file,
        sort_keys=False
    )

print("YAML configuration saved:", yaml_path)

# ================================================================
# 20. SAVE JSON RESULTS
# ================================================================

json_path = os.path.join(
    BASE_DIR,
    "electricity_consumption_results.json"
)

json_results = {

    "project_name": (
        "AI-Based Electricity Consumption "
        "Pattern Analysis"
    ),

    "generated_at": datetime.now().isoformat(),

    "total_sectors": int(len(df)),

    "total_consumption_MU": float(
        total_consumption
    ),

    "average_consumption_MU": float(
        average_consumption
    ),

    "rule_agreement_percentage": float(
        agreement
    ),

    "independent_test_accuracy": None,

    "class_distribution": {
        name: int(
            (df["Actual_Class"] == name).sum()
        )
        for name in CLASS_NAMES
    },

    "sector_results": result_df.to_dict(
        orient="records"
    )
}

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_results,
        file,
        indent=4,
        ensure_ascii=False
    )

print("JSON results saved:", json_path)

# ================================================================
# 21. SAVE PROCESSED DATA
# ================================================================

processed_path = os.path.join(
    BASE_DIR,
    "processed_electricity_consumption.csv"
)

df.to_csv(
    processed_path,
    index=False,
    encoding="utf-8-sig"
)

print("Processed dataset saved:", processed_path)

# ================================================================
# 22. VERIFY SAVED MODEL
# ================================================================

with open(pkl_path, "rb") as file:

    loaded_model = pickle.load(file)

loaded_predictions = loaded_model.predict(
    df[["Consumption"]].values
)

assert np.array_equal(
    loaded_predictions,
    df["Predicted_Class"].values
)

print("\nSaved model verification successful!")

# ================================================================
# 23. FINAL OUTPUT SUMMARY
# ================================================================

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "confusion_matrix.png",
    "electricity_consumption_model.pkl",
    "electricity_consumption_model.h5",
    "electricity_consumption_config.yaml",
    "electricity_consumption_results.json",
    "processed_electricity_consumption.csv"
]

print("\n" + "=" * 75)
print("ALL OUTPUTS GENERATED")
print("=" * 75)

for filename in output_files:

    full_path = os.path.join(
        BASE_DIR,
        filename
    )

    if os.path.exists(full_path):

        print("[SAVED]", filename)

    else:

        print("[MISSING]", filename)

print("\nOutput directory:")
print(BASE_DIR)

print("\nIMPORTANT:")
print(
    "The dataset is too small for meaningful "
    "supervised ML validation."
)

print(
    "The accuracy graph represents agreement "
    "with rule-generated labels, not "
    "independent predictive accuracy."
)

print("\nPROJECT EXECUTION COMPLETED!")
```
