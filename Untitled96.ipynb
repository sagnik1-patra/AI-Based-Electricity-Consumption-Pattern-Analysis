```python
import os
import json
import pickle
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

warnings.filterwarnings("ignore")

# ================================================================
# 1. PROJECT CONFIGURATION
# ================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Electricity Consumption Pattern Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "Consumptionofelectricity_0.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)

# PSO hyperparameters
NUM_PARTICLES = 40
MAX_ITERATIONS = 100

INERTIA_WEIGHT = 0.7
COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

CLASS_NAMES = ["Low", "Medium", "High"]

CLASS_COLORS = {
    "Low": "#2ecc71",
    "Medium": "#f39c12",
    "High": "#e74c3c"
}

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

print("=" * 75)
print("ELECTRICITY CONSUMPTION ANALYSIS USING PSO")
print("=" * 75)

# ================================================================
# 2. OUTPUT HELPER FUNCTIONS
# ================================================================

def output_path(filename):
    return os.path.join(BASE_DIR, "pso_" + filename)


def save_show(filename):
    path = output_path(filename)

    plt.tight_layout()
    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

    print("[SAVED]", path)


def json_safe(value):

    if isinstance(value, dict):
        return {
            str(k): json_safe(v)
            for k, v in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [json_safe(v) for v in value]

    if isinstance(value, np.ndarray):
        return value.tolist()

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    if isinstance(value, np.bool_):
        return bool(value)

    return value


# ================================================================
# 3. LOAD DATASET
# ================================================================

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\nDataset loaded successfully!")
print("Original shape:", df.shape)

print("\nOriginal columns:")
print(df.columns.tolist())

print("\nFirst five rows:")
print(df.head())

# ================================================================
# 4. CLEAN COLUMN NAMES
# ================================================================

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
)

rename_mapping = {}

for col in df.columns:

    normalized = (
        col.lower()
        .replace(" ", "")
        .replace("_", "")
    )

    if "category" in normalized or "sector" in normalized:
        rename_mapping[col] = "Category"

    elif "percentage" in normalized:
        rename_mapping[col] = "Percentage"

    elif "consumption" in normalized:
        rename_mapping[col] = "Consumption"

df.rename(
    columns=rename_mapping,
    inplace=True
)

required_columns = [
    "Category",
    "Consumption"
]

for col in required_columns:

    if col not in df.columns:
        raise ValueError(
            f"Missing required column: {col}"
        )

# ================================================================
# 5. DATA CLEANING
# ================================================================

df["Category"] = (
    df["Category"]
    .astype(str)
    .str.strip()
)

df["Consumption"] = pd.to_numeric(
    df["Consumption"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace(
        r"[^0-9.\-]",
        "",
        regex=True
    ),
    errors="coerce"
)

df.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)

df.dropna(
    subset=["Category", "Consumption"],
    inplace=True
)

df = df[
    (df["Category"] != "") &
    (df["Category"].str.lower() != "nan") &
    (df["Consumption"] >= 0)
].copy()

# Remove aggregate total rows
aggregate_names = {
    "total",
    "grand total",
    "all sectors",
    "overall total"
}

df = df[
    ~df["Category"]
    .str.lower()
    .isin(aggregate_names)
].copy()

# Combine duplicate sector categories
df = (
    df.groupby(
        "Category",
        as_index=False
    )["Consumption"]
    .sum()
)

if len(df) < 3:
    raise ValueError(
        "At least 3 valid sector records are required."
    )

df = df.sort_values(
    "Consumption",
    ascending=False
).reset_index(drop=True)

print("\nCleaned dataset:")
print(df)

# ================================================================
# 6. FEATURE ENGINEERING
# ================================================================

total_consumption = float(
    df["Consumption"].sum()
)

if total_consumption <= 0:
    raise ValueError(
        "Total electricity consumption must be positive."
    )

df["Percentage"] = (
    df["Consumption"] /
    total_consumption
) * 100

df["Consumption_Share"] = (
    df["Consumption"] /
    total_consumption
)

df["Normalized_Consumption"] = (
    df["Consumption"] /
    df["Consumption"].max()
)

df["Log_Consumption"] = np.log1p(
    df["Consumption"]
)

df["Consumption_Rank"] = (
    df["Consumption"]
    .rank(
        ascending=False,
        method="dense"
    )
    .astype(int)
)

# ================================================================
# 7. NORMALIZE CONSUMPTION FOR PSO
# ================================================================

consumption_values = (
    df["Consumption"]
    .to_numpy(dtype=float)
)

consumption_min = float(
    consumption_values.min()
)

consumption_max = float(
    consumption_values.max()
)

consumption_range = (
    consumption_max - consumption_min
)

if np.isclose(consumption_range, 0):
    raise ValueError(
        "All consumption values are identical. "
        "Three meaningful groups cannot be formed."
    )

X_normalized = (
    (consumption_values - consumption_min) /
    consumption_range
)

df["PSO_Normalized"] = X_normalized

print("\nNormalized consumption:")
print(X_normalized)

# ================================================================
# 8. PSO CLASSIFICATION FUNCTION
# ================================================================

# Each particle represents two thresholds:
#
# Particle = [threshold_1, threshold_2]
#
# Low    : x <= threshold_1
# Medium : threshold_1 < x <= threshold_2
# High   : x > threshold_2

def assign_clusters(values, thresholds):

    t1, t2 = np.sort(thresholds)

    labels = np.where(
        values <= t1,
        0,
        np.where(
            values <= t2,
            1,
            2
        )
    )

    return labels


# ================================================================
# 9. PSO OBJECTIVE FUNCTION
# ================================================================

# Minimize within-cluster sum of squared errors (SSE).
#
# Empty classes receive a large penalty.
#
# This is an unsupervised clustering objective,
# not supervised classification accuracy.

def objective_function(position):

    labels = assign_clusters(
        X_normalized,
        position
    )

    if len(np.unique(labels)) < 3:
        return 1e9

    total_sse = 0.0

    for cluster_id in range(3):

        cluster_values = X_normalized[
            labels == cluster_id
        ]

        center = cluster_values.mean()

        total_sse += np.sum(
            (cluster_values - center) ** 2
        )

    return float(total_sse)


# ================================================================
# 10. INITIALIZE PARTICLE SWARM
# ================================================================

positions = rng.uniform(
    low=0.0,
    high=1.0,
    size=(NUM_PARTICLES, 2)
)

positions = np.sort(
    positions,
    axis=1
)

velocities = rng.uniform(
    low=-0.1,
    high=0.1,
    size=(NUM_PARTICLES, 2)
)

personal_best_positions = (
    positions.copy()
)

personal_best_scores = np.array([
    objective_function(position)
    for position in positions
])

global_best_index = np.argmin(
    personal_best_scores
)

global_best_position = (
    personal_best_positions[
        global_best_index
    ].copy()
)

global_best_score = float(
    personal_best_scores[
        global_best_index
    ]
)

fitness_history = []

print("\nStarting PSO optimization...")

# ================================================================
# 11. PARTICLE SWARM OPTIMIZATION
# ================================================================

for iteration in range(MAX_ITERATIONS):

    for i in range(NUM_PARTICLES):

        r1 = rng.random(2)
        r2 = rng.random(2)

        cognitive_velocity = (
            COGNITIVE_COEFFICIENT *
            r1 *
            (
                personal_best_positions[i] -
                positions[i]
            )
        )

        social_velocity = (
            SOCIAL_COEFFICIENT *
            r2 *
            (
                global_best_position -
                positions[i]
            )
        )

        velocities[i] = (
            INERTIA_WEIGHT * velocities[i] +
            cognitive_velocity +
            social_velocity
        )

        # Prevent excessively large movements
        velocities[i] = np.clip(
            velocities[i],
            -0.2,
            0.2
        )

        positions[i] += velocities[i]

        positions[i] = np.clip(
            positions[i],
            0.0,
            1.0
        )

        positions[i] = np.sort(
            positions[i]
        )

        current_score = objective_function(
            positions[i]
        )

        if (
            current_score <
            personal_best_scores[i]
        ):

            personal_best_scores[i] = (
                current_score
            )

            personal_best_positions[i] = (
                positions[i].copy()
            )

        if current_score < global_best_score:

            global_best_score = float(
                current_score
            )

            global_best_position = (
                positions[i].copy()
            )

    fitness_history.append(
        global_best_score
    )

    if (
        iteration == 0 or
        (iteration + 1) % 10 == 0
    ):

        print(
            f"Iteration {iteration + 1:3d} "
            f"| Best SSE: {global_best_score:.8f}"
        )

# ================================================================
# 12. VERIFY OPTIMIZATION RESULT
# ================================================================

if global_best_score >= 1e8:
    raise RuntimeError(
        "PSO could not identify three non-empty groups."
    )

best_thresholds = np.sort(
    global_best_position
)

print("\nPSO optimization completed!")
print("Best normalized thresholds:", best_thresholds)
print("Best SSE:", global_best_score)

# ================================================================
# 13. CONVERT THRESHOLDS TO ORIGINAL UNITS
# ================================================================

threshold_1_MU = float(
    consumption_min +
    best_thresholds[0] * consumption_range
)

threshold_2_MU = float(
    consumption_min +
    best_thresholds[1] * consumption_range
)

print("\nOptimized PSO thresholds:")
print(
    f"Low/Medium threshold: "
    f"{threshold_1_MU:,.4f} M.U."
)

print(
    f"Medium/High threshold: "
    f"{threshold_2_MU:,.4f} M.U."
)

# ================================================================
# 14. ASSIGN PSO CLASSES
# ================================================================

cluster_ids = assign_clusters(
    X_normalized,
    best_thresholds
)

df["PSO_Class_Code"] = cluster_ids

df["PSO_Class"] = [
    CLASS_NAMES[index]
    for index in cluster_ids
]

# ================================================================
# 15. CREATE REUSABLE PSO MODEL
# ================================================================

pso_model = {

    "model_type": (
        "PSO_Optimized_Threshold_Classifier"
    ),

    "threshold_1_MU": threshold_1_MU,

    "threshold_2_MU": threshold_2_MU,

    "normalized_thresholds": (
        best_thresholds.tolist()
    ),

    "best_sse": global_best_score,

    "classes": CLASS_NAMES,

    "random_state": RANDOM_STATE,

    "num_particles": NUM_PARTICLES,

    "max_iterations": MAX_ITERATIONS
}


def pso_predict(values, model):

    values = np.asarray(
        values,
        dtype=float
    )

    t1 = model["threshold_1_MU"]
    t2 = model["threshold_2_MU"]

    return np.where(
        values <= t1,
        "Low",
        np.where(
            values <= t2,
            "Medium",
            "High"
        )
    )


df["PSO_Predicted_Class"] = pso_predict(
    df["Consumption"].to_numpy(),
    pso_model
)

df["PSO_Predicted_Code"] = (
    df["PSO_Predicted_Class"]
    .map({
        "Low": 0,
        "Medium": 1,
        "High": 2
    })
    .astype(int)
)

print("\nPSO Classification Results:")

print(
    df[
        [
            "Category",
            "Consumption",
            "Percentage",
            "PSO_Class",
            "PSO_Predicted_Class"
        ]
    ].to_string(index=False)
)

# ================================================================
# 16. CALCULATE RULE AGREEMENT
# ================================================================

# PSO-generated labels and predictions are based
# on the same thresholds.
#
# Agreement is NOT independent predictive accuracy.

df["PSO_Rule_Match"] = (
    df["PSO_Class"] ==
    df["PSO_Predicted_Class"]
)

overall_agreement = float(
    df["PSO_Rule_Match"].mean() * 100
)

class_agreement = {}

for class_name in CLASS_NAMES:

    subset = df[
        df["PSO_Class"] == class_name
    ]

    if len(subset) > 0:

        class_agreement[class_name] = float(
            subset["PSO_Rule_Match"].mean() * 100
        )

    else:

        class_agreement[class_name] = 0.0

print(
    "\nPSO rule agreement:",
    f"{overall_agreement:.2f}%"
)

print(
    "This is not independent ML test accuracy."
)

# ================================================================
# 17. PSO ACCURACY GRAPH
# ================================================================

plt.figure(figsize=(10, 6))

graph_classes = CLASS_NAMES + ["Overall"]

graph_values = [
    class_agreement["Low"],
    class_agreement["Medium"],
    class_agreement["High"],
    overall_agreement
]

graph_colors = [
    CLASS_COLORS["Low"],
    CLASS_COLORS["Medium"],
    CLASS_COLORS["High"],
    "#3498db"
]

bars = plt.bar(
    graph_classes,
    graph_values,
    color=graph_colors,
    edgecolor="black"
)

for bar, value in zip(bars, graph_values):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 1,
        f"{value:.2f}%",
        ha="center",
        fontweight="bold"
    )

plt.title(
    "PSO Rule-Based Classification Agreement\n"
    "(Not Independent Test Accuracy)",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Electricity Consumption Class")
plt.ylabel("Agreement (%)")

plt.ylim(0, 110)

save_show("accuracy_graph.png")

# ================================================================
# 18. PSO CORRELATION HEATMAP
# ================================================================

numeric_features = [
    "Consumption",
    "Percentage",
    "Consumption_Share",
    "Normalized_Consumption",
    "Log_Consumption",
    "Consumption_Rank",
    "PSO_Normalized",
    "PSO_Class_Code"
]

correlation_matrix = (
    df[numeric_features].corr()
)

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title(
    "PSO Electricity Consumption Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

save_show("heatmap.png")

# ================================================================
# 19. PSO COMPARISON GRAPH
# ================================================================

average_consumption = float(
    df["Consumption"].mean()
)

plt.figure(figsize=(13, 7))

bars = plt.bar(
    df["Category"],
    df["Consumption"],
    color=[
        CLASS_COLORS[level]
        for level in df["PSO_Class"]
    ],
    edgecolor="black"
)

plt.axhline(
    average_consumption,
    color="blue",
    linestyle="--",
    linewidth=2,
    label=(
        f"Average Consumption: "
        f"{average_consumption:,.2f} M.U."
    )
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom"
    )

plt.title(
    "PSO Sector-Wise Electricity Consumption Comparison",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Electricity Consumption Sector")
plt.ylabel("Consumption (Million Units)")

plt.xticks(
    rotation=35,
    ha="right"
)

plt.legend()

save_show("comparison_graph.png")

# ================================================================
# 20. SAVE PSO RESULT CSV
# ================================================================

result_df = df[
    [
        "Category",
        "Consumption",
        "Percentage",
        "Consumption_Rank",
        "PSO_Normalized",
        "PSO_Class",
        "PSO_Class_Code",
        "PSO_Predicted_Class",
        "PSO_Rule_Match"
    ]
].copy()

result_csv_path = output_path(
    "result.csv"
)

result_df.to_csv(
    result_csv_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nPSO Result CSV:")
print(result_df)

print("[SAVED]", result_csv_path)

# ================================================================
# 21. PSO RESULT GRAPH
# ================================================================

class_consumption = (
    df.groupby("PSO_Class")["Consumption"]
    .sum()
    .reindex(CLASS_NAMES, fill_value=0)
)

plt.figure(figsize=(10, 7))

bars = plt.bar(
    class_consumption.index,
    class_consumption.values,
    color=[
        CLASS_COLORS[name]
        for name in class_consumption.index
    ],
    edgecolor="black"
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.title(
    "PSO Electricity Consumption Classification Results",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("PSO Consumption Class")
plt.ylabel("Total Consumption (M.U.)")

save_show("result_graph.png")

# ================================================================
# 22. SAVE PSO PREDICTION CSV
# ================================================================

prediction_df = pd.DataFrame({

    "Category": df["Category"],

    "Consumption_MU": (
        df["Consumption"]
    ),

    "PSO_Predicted_Class": (
        df["PSO_Predicted_Class"]
    ),

    "PSO_Predicted_Code": (
        df["PSO_Predicted_Code"]
    ),

    "Consumption_Percentage": (
        df["Percentage"]
    )
})

prediction_csv_path = output_path(
    "prediction.csv"
)

prediction_df.to_csv(
    prediction_csv_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nPSO Prediction CSV:")
print(prediction_df)

print("[SAVED]", prediction_csv_path)

# ================================================================
# 23. PSO PREDICTION GRAPH
# ================================================================

plt.figure(figsize=(13, 7))

bars = plt.bar(
    prediction_df["Category"],
    prediction_df["Consumption_MU"],
    color=[
        CLASS_COLORS[level]
        for level in
        prediction_df["PSO_Predicted_Class"]
    ],
    edgecolor="black"
)

for bar, level in zip(
    bars,
    prediction_df["PSO_Predicted_Class"]
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        level,
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.title(
    "PSO Predicted Electricity Consumption Classes",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Electricity Consumption Sector")
plt.ylabel("Consumption (Million Units)")

plt.xticks(
    rotation=35,
    ha="right"
)

save_show("prediction_graph.png")

# ================================================================
# 24. PSO CONFUSION MATRIX
# ================================================================

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    df["PSO_Class"],
    df["PSO_Predicted_Class"],
    labels=CLASS_NAMES
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.title(
    "PSO Rule Agreement Confusion Matrix\n"
    "(Not Independent Validation)",
    fontsize=13
)

plt.xlabel("Predicted Class")
plt.ylabel("PSO-Generated Class")

save_show("confusion_matrix.png")

# ================================================================
# 25. PSO FITNESS GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    range(1, MAX_ITERATIONS + 1),
    fitness_history,
    color="purple",
    linewidth=2,
    label="Global Best SSE"
)

plt.xlabel("Iteration")
plt.ylabel("Within-Class SSE")

plt.title(
    "PSO Optimization Convergence",
    fontsize=14,
    fontweight="bold"
)

plt.legend()

save_show("fitness_graph.png")

# ================================================================
# 26. PSO CLASS DISTRIBUTION GRAPH
# ================================================================

class_counts = (
    df["PSO_Class"]
    .value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
)

plt.figure(figsize=(9, 6))

bars = plt.bar(
    class_counts.index,
    class_counts.values,
    color=[
        CLASS_COLORS[name]
        for name in class_counts.index
    ],
    edgecolor="black"
)

for bar in bars:

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.05,
        str(int(bar.get_height())),
        ha="center",
        fontweight="bold"
    )

plt.title(
    "PSO Electricity Consumption Class Distribution",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("PSO Consumption Class")
plt.ylabel("Number of Sectors")

plt.ylim(
    0,
    max(class_counts.max() + 1, 2)
)

save_show("class_distribution_graph.png")

# ================================================================
# 27. SAVE PSO MODEL AS PKL
# ================================================================

pkl_path = output_path(
    "model.pkl"
)

with open(pkl_path, "wb") as file:

    pickle.dump(pso_model, file)

print("[SAVED]", pkl_path)

# ================================================================
# 28. SAVE PSO MODEL AS H5
# ================================================================

h5_path = output_path(
    "model.h5"
)

with h5py.File(h5_path, "w") as h5file:

    h5file.attrs["model_type"] = (
        "PSO Optimized Threshold Classifier"
    )

    h5file.attrs["created_at"] = (
        datetime.now().isoformat()
    )

    parameters = h5file.create_group(
        "model_parameters"
    )

    parameters.create_dataset(
        "best_normalized_thresholds",
        data=best_thresholds
    )

    parameters.create_dataset(
        "threshold_1_MU",
        data=threshold_1_MU
    )

    parameters.create_dataset(
        "threshold_2_MU",
        data=threshold_2_MU
    )

    parameters.create_dataset(
        "best_sse",
        data=global_best_score
    )

    parameters.create_dataset(
        "fitness_history",
        data=np.asarray(fitness_history)
    )

    data_group = h5file.create_group(
        "processed_data"
    )

    string_dtype = h5py.string_dtype(
        encoding="utf-8"
    )

    data_group.create_dataset(
        "categories",
        data=df["Category"].tolist(),
        dtype=string_dtype
    )

    data_group.create_dataset(
        "consumption",
        data=df["Consumption"].to_numpy()
    )

    data_group.create_dataset(
        "predicted_classes",
        data=df["PSO_Predicted_Class"].tolist(),
        dtype=string_dtype
    )

    data_group.create_dataset(
        "class_codes",
        data=df["PSO_Predicted_Code"].to_numpy()
    )

print("[SAVED]", h5_path)

# ================================================================
# 29. SAVE PSO YAML CONFIGURATION
# ================================================================

yaml_config = {

    "project_name": (
        "AI-Based Electricity Consumption "
        "Pattern Analysis Using PSO"
    ),

    "dataset_path": DATASET_PATH,

    "algorithm": {
        "name": "Particle Swarm Optimization",
        "num_particles": NUM_PARTICLES,
        "max_iterations": MAX_ITERATIONS,
        "inertia_weight": INERTIA_WEIGHT,
        "cognitive_coefficient": COGNITIVE_COEFFICIENT,
        "social_coefficient": SOCIAL_COEFFICIENT,
        "random_state": RANDOM_STATE
    },

    "classification": {
        "classes": CLASS_NAMES,
        "threshold_1_MU": threshold_1_MU,
        "threshold_2_MU": threshold_2_MU
    },

    "optimization": {
        "objective": "Minimize Within-Class SSE",
        "best_sse": global_best_score
    },

    "evaluation": {
        "rule_agreement_percentage": (
            overall_agreement
        ),
        "independent_test_accuracy": None
    },

    "output_directory": BASE_DIR
}

yaml_path = output_path(
    "config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_config,
        file,
        sort_keys=False
    )

print("[SAVED]", yaml_path)

# ================================================================
# 30. SAVE PSO JSON RESULTS
# ================================================================

json_results = {

    "project_name": (
        "AI-Based Electricity Consumption "
        "Pattern Analysis Using PSO"
    ),

    "generated_at": datetime.now().isoformat(),

    "algorithm": "Particle Swarm Optimization",

    "total_sectors": int(len(df)),

    "total_consumption_MU": float(
        total_consumption
    ),

    "average_consumption_MU": float(
        average_consumption
    ),

    "best_normalized_thresholds": (
        best_thresholds.tolist()
    ),

    "best_sse": global_best_score,

    "threshold_1_MU": threshold_1_MU,

    "threshold_2_MU": threshold_2_MU,

    "rule_agreement_percentage": (
        overall_agreement
    ),

    "independent_test_accuracy": None,

    "class_distribution": {
        name: int(class_counts[name])
        for name in CLASS_NAMES
    },

    "fitness_history": fitness_history,

    "sector_results": result_df.to_dict(
        orient="records"
    )
}

json_path = output_path(
    "results.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_safe(json_results),
        file,
        indent=4,
        ensure_ascii=False
    )

print("[SAVED]", json_path)

# ================================================================
# 31. SAVE PROCESSED DATA
# ================================================================

processed_path = output_path(
    "processed_data.csv"
)

df.to_csv(
    processed_path,
    index=False,
    encoding="utf-8-sig"
)

print("[SAVED]", processed_path)

# ================================================================
# 32. VERIFY SAVED PSO MODEL
# ================================================================

with open(pkl_path, "rb") as file:

    loaded_model = pickle.load(file)

loaded_predictions = pso_predict(
    df["Consumption"].to_numpy(),
    loaded_model
)

assert np.array_equal(
    loaded_predictions,
    df["PSO_Predicted_Class"].to_numpy()
)

print("\nPSO model verification successful!")

# ================================================================
# 33. FINAL OUTPUT VERIFICATION
# ================================================================

expected_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "confusion_matrix.png",
    "fitness_graph.png",
    "class_distribution_graph.png",
    "model.pkl",
    "model.h5",
    "config.yaml",
    "results.json",
    "processed_data.csv"
]

print("\n" + "=" * 75)
print("PSO PROJECT OUTPUT SUMMARY")
print("=" * 75)

for filename in expected_files:

    path = output_path(filename)

    if os.path.exists(path):
        print("[SAVED]", os.path.basename(path))

    else:
        print("[MISSING]", os.path.basename(path))

print("\nTotal expected files:", len(expected_files))

print("\nAll outputs saved in:")
print(BASE_DIR)

print("\nPSO Best SSE:", global_best_score)

print(
    "PSO Rule Agreement:",
    f"{overall_agreement:.2f}%"
)

print(
    "\nNOTE: Rule agreement is not "
    "independent predictive accuracy."
)

print("\nPSO EXECUTION COMPLETED SUCCESSFULLY!")
```
