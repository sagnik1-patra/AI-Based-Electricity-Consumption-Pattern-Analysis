```python
import os
import json
import pickle
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

warnings.filterwarnings("ignore")

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix

# ================================================================
# 1. CONFIGURATION
# ================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Electricity Consumption Pattern Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "Consumptionofelectricity_0.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)

POPULATION_SIZE = 40
GENERATIONS = 100
CLONE_FACTOR = 5
MUTATION_RATE = 0.12
ELITE_COUNT = 8

CLASS_NAMES = ["Low", "Medium", "High"]

CLASS_COLORS = {
    "Low": "#2ecc71",
    "Medium": "#f39c12",
    "High": "#e74c3c"
}

sns.set_theme(style="whitegrid")

print("=" * 75)
print("ELECTRICITY CONSUMPTION ANALYSIS USING AIS")
print("=" * 75)

# ================================================================
# 2. OUTPUT HELPER FUNCTIONS
# ================================================================

def output_path(filename):
    return os.path.join(BASE_DIR, "ais_" + filename)


def save_show(filename):
    path = output_path(filename)

    plt.tight_layout()
    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

    print("[SAVED]", path)


def json_safe(value):
    if isinstance(value, dict):
        return {
            str(k): json_safe(v)
            for k, v in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [json_safe(v) for v in value]

    if isinstance(value, np.ndarray):
        return value.tolist()

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    if isinstance(value, np.bool_):
        return bool(value)

    return value


# ================================================================
# 3. LOAD DATASET
# ================================================================

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(
    DATASET_PATH,
    encoding="utf-8-sig"
)

print("\nOriginal dataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print(df.head())

# ================================================================
# 4. CLEAN COLUMN NAMES
# ================================================================

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
)

rename_mapping = {}

for col in df.columns:

    normalized = (
        col.lower()
        .replace(" ", "")
        .replace("_", "")
    )

    if "category" in normalized or "sector" in normalized:
        rename_mapping[col] = "Category"

    elif "percentage" in normalized:
        rename_mapping[col] = "Percentage"

    elif "consumption" in normalized:
        rename_mapping[col] = "Consumption"

df.rename(
    columns=rename_mapping,
    inplace=True
)

required_columns = ["Category", "Consumption"]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(
            f"Missing required column: {col}"
        )

# ================================================================
# 5. DATA CLEANING
# ================================================================

df["Category"] = (
    df["Category"]
    .astype(str)
    .str.strip()
)

df["Consumption"] = pd.to_numeric(
    df["Consumption"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace(
        r"[^0-9.\-]",
        "",
        regex=True
    ),
    errors="coerce"
)

df.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)

df.dropna(
    subset=["Category", "Consumption"],
    inplace=True
)

df = df[
    (df["Category"] != "") &
    (df["Category"].str.lower() != "nan") &
    (df["Consumption"] >= 0)
].copy()

# Remove aggregate totals
aggregate_names = {
    "total",
    "grand total",
    "all sectors",
    "overall total"
}

df = df[
    ~df["Category"]
    .str.lower()
    .isin(aggregate_names)
].copy()

# Aggregate duplicate sectors
df = (
    df.groupby(
        "Category",
        as_index=False
    )["Consumption"]
    .sum()
)

if len(df) < 3:
    raise ValueError(
        "At least 3 valid sector records are required."
    )

df = df.sort_values(
    "Consumption",
    ascending=False
).reset_index(drop=True)

print("\nCleaned dataset:")
print(df)

# ================================================================
# 6. FEATURE ENGINEERING
# ================================================================

total_consumption = df["Consumption"].sum()

if total_consumption <= 0:
    raise ValueError(
        "Total electricity consumption must be positive."
    )

df["Percentage"] = (
    df["Consumption"] / total_consumption
) * 100

df["Consumption_Share"] = (
    df["Consumption"] / total_consumption
)

df["Normalized_Consumption"] = (
    df["Consumption"] /
    df["Consumption"].max()
)

df["Log_Consumption"] = np.log1p(
    df["Consumption"]
)

df["Consumption_Rank"] = (
    df["Consumption"]
    .rank(
        ascending=False,
        method="dense"
    )
    .astype(int)
)

# ================================================================
# 7. STANDARDIZE CONSUMPTION
# ================================================================

scaler = StandardScaler()

X = df[["Consumption"]].to_numpy()

X_scaled = scaler.fit_transform(X).ravel()

# Normalize to [0, 1] for AIS optimization
x_min = float(X_scaled.min())
x_max = float(X_scaled.max())

if np.isclose(x_min, x_max):
    raise ValueError(
        "All sectors have identical consumption. "
        "AIS cannot identify meaningful groups."
    )

X_normalized = (
    (X_scaled - x_min) /
    (x_max - x_min)
)

df["AIS_Normalized"] = X_normalized

# ================================================================
# 8. AIS FITNESS FUNCTION
# ================================================================

# Each antibody represents two ordered thresholds.
#
# [threshold_1, threshold_2]
#
# Classes:
# Low    : x <= threshold_1
# Medium : threshold_1 < x <= threshold_2
# High   : x > threshold_2
#
# Fitness minimizes within-class squared error.
# Empty classes are heavily penalized.

def assign_clusters(values, antibody):

    t1, t2 = sorted(antibody)

    return np.where(
        values <= t1,
        0,
        np.where(
            values <= t2,
            1,
            2
        )
    )


def fitness_function(antibody, values):

    labels = assign_clusters(
        values,
        antibody
    )

    # Penalize empty clusters
    if len(np.unique(labels)) < 3:
        return -1e9

    total_sse = 0.0

    for cluster_id in range(3):

        cluster_values = values[
            labels == cluster_id
        ]

        center = cluster_values.mean()

        total_sse += np.sum(
            (cluster_values - center) ** 2
        )

    return -float(total_sse)


# ================================================================
# 9. INITIALIZE AIS POPULATION
# ================================================================

def create_antibody():
    return np.sort(
        rng.uniform(
            0.0,
            1.0,
            size=2
        )
    )


population = np.array([
    create_antibody()
    for _ in range(POPULATION_SIZE)
])

best_antibody = None
best_fitness = -np.inf

fitness_history = []

print("\nStarting AIS optimization...")

# ================================================================
# 10. AIS CLONAL SELECTION OPTIMIZATION
# ================================================================

for generation in range(GENERATIONS):

    fitness_scores = np.array([
        fitness_function(
            antibody,
            X_normalized
        )
        for antibody in population
    ])

    ranked_indices = np.argsort(
        fitness_scores
    )[::-1]

    population = population[
        ranked_indices
    ]

    fitness_scores = fitness_scores[
        ranked_indices
    ]

    if fitness_scores[0] > best_fitness:

        best_fitness = float(
            fitness_scores[0]
        )

        best_antibody = (
            population[0].copy()
        )

    fitness_history.append(
        best_fitness
    )

    # Select best antibodies
    elites = population[
        :ELITE_COUNT
    ].copy()

    clones = []

    for rank, antibody in enumerate(elites):

        # Better antibodies receive more clones
        number_of_clones = max(
            1,
            CLONE_FACTOR * (
                ELITE_COUNT - rank
            )
        )

        # Better antibodies mutate less
        adaptive_mutation = (
            MUTATION_RATE *
            (rank + 1) /
            ELITE_COUNT
        )

        for _ in range(number_of_clones):

            clone = antibody.copy()

            mutation = rng.normal(
                0,
                adaptive_mutation,
                size=2
            )

            clone += mutation

            clone = np.clip(
                clone,
                0,
                1
            )

            clone = np.sort(clone)

            clones.append(clone)

    clones = np.array(clones)

    combined_population = np.vstack([
        population,
        clones
    ])

    combined_fitness = np.array([
        fitness_function(
            antibody,
            X_normalized
        )
        for antibody in combined_population
    ])

    sorted_indices = np.argsort(
        combined_fitness
    )[::-1]

    # Preserve strong antibodies
    survivors = combined_population[
        sorted_indices[
            :POPULATION_SIZE - 5
        ]
    ]

    # Introduce new antibodies for diversity
    new_antibodies = np.array([
        create_antibody()
        for _ in range(5)
    ])

    population = np.vstack([
        survivors,
        new_antibodies
    ])

    if (
        generation == 0 or
        (generation + 1) % 10 == 0
    ):

        print(
            f"Generation {generation + 1:3d} "
            f"| Best Fitness: {best_fitness:.6f}"
        )

if best_antibody is None or best_fitness <= -1e8:
    raise RuntimeError(
        "AIS failed to find three non-empty classes."
    )

print("\nAIS optimization completed!")
print("Best antibody:", best_antibody)
print("Best fitness:", best_fitness)

# ================================================================
# 11. GENERATE AIS CLASSIFICATIONS
# ================================================================

cluster_ids = assign_clusters(
    X_normalized,
    best_antibody
)

df["AIS_Class_Code"] = cluster_ids

df["AIS_Class"] = [
    CLASS_NAMES[index]
    for index in cluster_ids
]

# Convert normalized thresholds to original M.U.
normalized_consumption = (
    df["Consumption"] -
    df["Consumption"].min()
) / (
    df["Consumption"].max() -
    df["Consumption"].min()
)

consumption_min = float(
    df["Consumption"].min()
)

consumption_max = float(
    df["Consumption"].max()
)

threshold_1_MU = (
    consumption_min +
    best_antibody[0] *
    (consumption_max - consumption_min)
)

threshold_2_MU = (
    consumption_min +
    best_antibody[1] *
    (consumption_max - consumption_min)
)

print("\nAIS thresholds (M.U.):")
print("Low/Medium:", threshold_1_MU)
print("Medium/High:", threshold_2_MU)

print("\nAIS classification results:")
print(
    df[
        [
            "Category",
            "Consumption",
            "Percentage",
            "AIS_Class"
        ]
    ].to_string(index=False)
)

# ================================================================
# 12. DEFINE REUSABLE AIS CLASSIFIER
# ================================================================

# A dictionary is used as the portable model artifact.
# It avoids requiring a custom class definition when
# loading the pickle in another Python script.

ais_model = {
    "model_type": "AIS_Clonal_Selection_Threshold_Model",
    "threshold_1_MU": float(threshold_1_MU),
    "threshold_2_MU": float(threshold_2_MU),
    "best_antibody": best_antibody.tolist(),
    "best_fitness": float(best_fitness),
    "classes": CLASS_NAMES,
    "random_state": RANDOM_STATE
}


def ais_predict(consumption_values, model):

    values = np.asarray(
        consumption_values,
        dtype=float
    )

    t1 = model["threshold_1_MU"]
    t2 = model["threshold_2_MU"]

    return np.where(
        values <= t1,
        "Low",
        np.where(
            values <= t2,
            "Medium",
            "High"
        )
    )


df["AIS_Predicted_Class"] = ais_predict(
    df["Consumption"].to_numpy(),
    ais_model
)

df["AIS_Predicted_Code"] = (
    df["AIS_Predicted_Class"]
    .map({
        "Low": 0,
        "Medium": 1,
        "High": 2
    })
    .astype(int)
)

# ================================================================
# 13. CLASSIFICATION AGREEMENT
# ================================================================

# The labels are generated by the optimized thresholds.
# Comparing those labels with predictions from the same
# thresholds measures rule agreement, NOT independent accuracy.

df["AIS_Rule_Match"] = (
    df["AIS_Class"] ==
    df["AIS_Predicted_Class"]
)

overall_agreement = (
    df["AIS_Rule_Match"].mean()
) * 100

class_agreement = {}

for class_name in CLASS_NAMES:

    subset = df[
        df["AIS_Class"] == class_name
    ]

    if len(subset) > 0:

        class_agreement[class_name] = float(
            subset["AIS_Rule_Match"].mean()
            * 100
        )

    else:

        class_agreement[class_name] = 0.0

print(
    "\nAIS rule agreement:",
    f"{overall_agreement:.2f}%"
)

print(
    "This is not independent test accuracy."
)

# ================================================================
# 14. AIS ACCURACY GRAPH
# ================================================================

plt.figure(figsize=(10, 6))

graph_classes = CLASS_NAMES + ["Overall"]

graph_values = [
    class_agreement["Low"],
    class_agreement["Medium"],
    class_agreement["High"],
    overall_agreement
]

graph_colors = [
    CLASS_COLORS["Low"],
    CLASS_COLORS["Medium"],
    CLASS_COLORS["High"],
    "#3498db"
]

bars = plt.bar(
    graph_classes,
    graph_values,
    color=graph_colors,
    edgecolor="black"
)

for bar, value in zip(bars, graph_values):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 1,
        f"{value:.2f}%",
        ha="center",
        fontweight="bold"
    )

plt.title(
    "AIS Rule-Based Classification Agreement\n"
    "(Not Independent Test Accuracy)",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Consumption Classification")
plt.ylabel("Agreement (%)")
plt.ylim(0, 110)

save_show("accuracy_graph.png")

# ================================================================
# 15. AIS CORRELATION HEATMAP
# ================================================================

numeric_features = [
    "Consumption",
    "Percentage",
    "Consumption_Share",
    "Normalized_Consumption",
    "Log_Consumption",
    "Consumption_Rank",
    "AIS_Normalized",
    "AIS_Class_Code"
]

correlation_matrix = (
    df[numeric_features].corr()
)

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title(
    "AIS Electricity Consumption Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

save_show("heatmap.png")

# ================================================================
# 16. AIS COMPARISON GRAPH
# ================================================================

average_consumption = float(
    df["Consumption"].mean()
)

plt.figure(figsize=(13, 7))

bars = plt.bar(
    df["Category"],
    df["Consumption"],
    color=[
        CLASS_COLORS[level]
        for level in df["AIS_Class"]
    ],
    edgecolor="black"
)

plt.axhline(
    average_consumption,
    color="blue",
    linestyle="--",
    linewidth=2,
    label=(
        f"Mean Consumption: "
        f"{average_consumption:,.2f} M.U."
    )
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom"
    )

plt.title(
    "AIS Sector-Wise Electricity Consumption Comparison",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Electricity Sector")
plt.ylabel("Consumption (M.U.)")

plt.xticks(
    rotation=35,
    ha="right"
)

plt.legend()

save_show("comparison_graph.png")

# ================================================================
# 17. AIS RESULT CSV
# ================================================================

result_df = df[
    [
        "Category",
        "Consumption",
        "Percentage",
        "Consumption_Rank",
        "AIS_Normalized",
        "AIS_Class",
        "AIS_Class_Code",
        "AIS_Predicted_Class",
        "AIS_Rule_Match"
    ]
].copy()

result_csv_path = output_path(
    "result.csv"
)

result_df.to_csv(
    result_csv_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nAIS Result CSV:")
print(result_df)

print("[SAVED]", result_csv_path)

# ================================================================
# 18. AIS RESULT GRAPH
# ================================================================

class_consumption = (
    df.groupby("AIS_Class")["Consumption"]
    .sum()
    .reindex(CLASS_NAMES, fill_value=0)
)

plt.figure(figsize=(10, 7))

bars = plt.bar(
    class_consumption.index,
    class_consumption.values,
    color=[
        CLASS_COLORS[name]
        for name in class_consumption.index
    ],
    edgecolor="black"
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.title(
    "AIS Electricity Consumption Classification Results",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("AIS Consumption Class")
plt.ylabel("Total Consumption (M.U.)")

save_show("result_graph.png")

# ================================================================
# 19. AIS PREDICTION CSV
# ================================================================

prediction_df = pd.DataFrame({

    "Category": df["Category"],

    "Consumption_MU": (
        df["Consumption"]
    ),

    "AIS_Predicted_Class": (
        df["AIS_Predicted_Class"]
    ),

    "AIS_Predicted_Code": (
        df["AIS_Predicted_Code"]
    ),

    "Consumption_Percentage": (
        df["Percentage"]
    )

})

prediction_csv_path = output_path(
    "prediction.csv"
)

prediction_df.to_csv(
    prediction_csv_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nAIS Prediction CSV:")
print(prediction_df)

print("[SAVED]", prediction_csv_path)

# ================================================================
# 20. AIS PREDICTION GRAPH
# ================================================================

plt.figure(figsize=(13, 7))

bars = plt.bar(
    prediction_df["Category"],
    prediction_df["Consumption_MU"],
    color=[
        CLASS_COLORS[level]
        for level in
        prediction_df["AIS_Predicted_Class"]
    ],
    edgecolor="black"
)

for bar, level in zip(
    bars,
    prediction_df["AIS_Predicted_Class"]
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        level,
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.title(
    "AIS Predicted Electricity Consumption Classes",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Electricity Sector")
plt.ylabel("Consumption (M.U.)")

plt.xticks(
    rotation=35,
    ha="right"
)

save_show("prediction_graph.png")

# ================================================================
# 21. AIS CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    df["AIS_Class"],
    df["AIS_Predicted_Class"],
    labels=CLASS_NAMES
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.title(
    "AIS Rule Agreement Confusion Matrix\n"
    "(Not Independent Validation)",
    fontsize=13
)

plt.xlabel("Predicted Class")
plt.ylabel("AIS-Generated Class")

save_show("confusion_matrix.png")

# ================================================================
# 22. AIS FITNESS GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    range(1, GENERATIONS + 1),
    fitness_history,
    color="purple",
    linewidth=2,
    label="Best AIS Fitness"
)

plt.xlabel("Generation")
plt.ylabel("Fitness (Negative Within-Class SSE)")

plt.title(
    "AIS Clonal Selection Optimization Convergence",
    fontsize=14,
    fontweight="bold"
)

plt.legend()

save_show("fitness_graph.png")

# ================================================================
# 23. AIS CLASS DISTRIBUTION GRAPH
# ================================================================

class_counts = (
    df["AIS_Class"]
    .value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
)

plt.figure(figsize=(9, 6))

bars = plt.bar(
    class_counts.index,
    class_counts.values,
    color=[
        CLASS_COLORS[name]
        for name in class_counts.index
    ],
    edgecolor="black"
)

for bar in bars:

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.05,
        str(int(bar.get_height())),
        ha="center",
        fontweight="bold"
    )

plt.xlabel("AIS Consumption Class")
plt.ylabel("Number of Sectors")

plt.title(
    "AIS Electricity Consumption Class Distribution",
    fontsize=14,
    fontweight="bold"
)

plt.ylim(
    0,
    max(class_counts.max() + 1, 2)
)

save_show("class_distribution_graph.png")

# ================================================================
# 24. SAVE AIS MODEL AS PKL
# ================================================================

pkl_path = output_path(
    "model.pkl"
)

with open(pkl_path, "wb") as file:
    pickle.dump(ais_model, file)

print("[SAVED]", pkl_path)

# ================================================================
# 25. SAVE AIS MODEL AS H5
# ================================================================

h5_path = output_path(
    "model.h5"
)

with h5py.File(h5_path, "w") as h5file:

    h5file.attrs["model_type"] = (
        "AIS Clonal Selection Threshold Model"
    )

    h5file.attrs["created_at"] = (
        datetime.now().isoformat()
    )

    parameters = h5file.create_group(
        "model_parameters"
    )

    parameters.create_dataset(
        "best_antibody",
        data=best_antibody
    )

    parameters.create_dataset(
        "threshold_1_MU",
        data=threshold_1_MU
    )

    parameters.create_dataset(
        "threshold_2_MU",
        data=threshold_2_MU
    )

    parameters.create_dataset(
        "best_fitness",
        data=best_fitness
    )

    parameters.create_dataset(
        "fitness_history",
        data=np.asarray(fitness_history)
    )

    data_group = h5file.create_group(
        "processed_data"
    )

    string_dtype = h5py.string_dtype(
        encoding="utf-8"
    )

    data_group.create_dataset(
        "categories",
        data=df["Category"].tolist(),
        dtype=string_dtype
    )

    data_group.create_dataset(
        "consumption",
        data=df["Consumption"].to_numpy()
    )

    data_group.create_dataset(
        "predicted_classes",
        data=df["AIS_Predicted_Class"].tolist(),
        dtype=string_dtype
    )

    data_group.create_dataset(
        "class_codes",
        data=df["AIS_Predicted_Code"].to_numpy()
    )

print("[SAVED]", h5_path)

# ================================================================
# 26. SAVE AIS YAML CONFIGURATION
# ================================================================

yaml_config = {

    "project_name": (
        "AI-Based Electricity Consumption "
        "Pattern Analysis Using AIS"
    ),

    "dataset_path": DATASET_PATH,

    "algorithm": {
        "name": "Clonal Selection Inspired AIS",
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "clone_factor": CLONE_FACTOR,
        "mutation_rate": MUTATION_RATE,
        "elite_count": ELITE_COUNT,
        "random_state": RANDOM_STATE
    },

    "classification": {
        "classes": CLASS_NAMES,
        "threshold_1_MU": float(threshold_1_MU),
        "threshold_2_MU": float(threshold_2_MU)
    },

    "optimization": {
        "objective": "Minimize Within-Class SSE",
        "best_fitness": float(best_fitness)
    },

    "evaluation": {
        "rule_agreement_percentage": float(
            overall_agreement
        ),
        "independent_test_accuracy": None
    },

    "output_directory": BASE_DIR
}

yaml_path = output_path(
    "config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_config,
        file,
        sort_keys=False
    )

print("[SAVED]", yaml_path)

# ================================================================
# 27. SAVE AIS JSON RESULTS
# ================================================================

json_results = {

    "project_name": (
        "AI-Based Electricity Consumption "
        "Pattern Analysis Using AIS"
    ),

    "generated_at": datetime.now().isoformat(),

    "algorithm": "AIS Clonal Selection",

    "total_sectors": int(len(df)),

    "total_consumption_MU": float(
        total_consumption
    ),

    "average_consumption_MU": float(
        average_consumption
    ),

    "best_antibody": best_antibody.tolist(),

    "best_fitness": float(best_fitness),

    "threshold_1_MU": float(threshold_1_MU),

    "threshold_2_MU": float(threshold_2_MU),

    "rule_agreement_percentage": float(
        overall_agreement
    ),

    "independent_test_accuracy": None,

    "class_distribution": {
        name: int(class_counts[name])
        for name in CLASS_NAMES
    },

    "fitness_history": fitness_history,

    "sector_results": result_df.to_dict(
        orient="records"
    )
}

json_path = output_path(
    "results.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_safe(json_results),
        file,
        indent=4,
        ensure_ascii=False
    )

print("[SAVED]", json_path)

# ================================================================
# 28. SAVE PROCESSED DATA
# ================================================================

processed_path = output_path(
    "processed_data.csv"
)

df.to_csv(
    processed_path,
    index=False,
    encoding="utf-8-sig"
)

print("[SAVED]", processed_path)

# ================================================================
# 29. VERIFY SAVED AIS MODEL
# ================================================================

with open(pkl_path, "rb") as file:

    loaded_model = pickle.load(file)

loaded_predictions = ais_predict(
    df["Consumption"].to_numpy(),
    loaded_model
)

assert np.array_equal(
    loaded_predictions,
    df["AIS_Predicted_Class"].to_numpy()
)

print("\nAIS model verification successful!")

# ================================================================
# 30. FINAL OUTPUT VERIFICATION
# ================================================================

expected_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "confusion_matrix.png",
    "fitness_graph.png",
    "class_distribution_graph.png",
    "model.pkl",
    "model.h5",
    "config.yaml",
    "results.json",
    "processed_data.csv"
]

print("\n" + "=" * 75)
print("AIS PROJECT OUTPUT SUMMARY")
print("=" * 75)

for filename in expected_files:

    path = output_path(filename)

    if os.path.exists(path):
        print("[SAVED]", os.path.basename(path))

    else:
        print("[MISSING]", os.path.basename(path))

print("\nTotal expected files:", len(expected_files))

print("\nAll outputs saved in:")
print(BASE_DIR)

print("\nAIS Best Fitness:", best_fitness)

print(
    "AIS Rule Agreement:",
    f"{overall_agreement:.2f}%"
)

print(
    "\nNOTE: Rule agreement is not "
    "independent predictive accuracy."
)

print("\nAIS EXECUTION COMPLETED SUCCESSFULLY!")
```
